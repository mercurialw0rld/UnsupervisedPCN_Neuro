## 2. Utility Functions and Data Processing
We define functions to handle data binning, synthetic grating generation, and the construction of biological-inspired convolutional filters.

```python
import torch.nn as nn
import torch.optim as optim
import torch


def load_data(data_name, bin_width=1):
  """Load mouse V1 data from Stringer et al. (2019)

  Data from study reported in this preprint:
  https://www.biorxiv.org/content/10.1101/679324v2.abstract

  These data comprise time-averaged responses of ~20,000 neurons
  to ~4,000 stimulus gratings of different orientations, recorded
  through Calcium imaginge. The responses have been normalized by
  spontaneous levels of activity and then z-scored over stimuli, so
  expect negative numbers. They have also been binned and averaged
  to each degree of orientation.

  This function returns the relevant data (neural responses and
  stimulus orientations) in a torch.Tensor of data type torch.float32
  in order to match the default data type for nn.Parameters in
  Google Colab.

  This function will actually average responses to stimuli with orientations
  falling within bins specified by the bin_width argument. This helps
  produce individual neural "responses" with smoother and more
  interpretable tuning curves.

  Args:
    bin_width (float): size of stimulus bins over which to average neural
      responses

  Returns:
    resp (torch.Tensor): n_stimuli x n_neurons matrix of neural responses,
        each row contains the responses of each neuron to a given stimulus.
        As mentioned above, neural "response" is actually an average over
        responses to stimuli with similar angles falling within specified bins.
    stimuli: (torch.Tensor): n_stimuli x 1 column vector with orientation
        of each stimulus, in degrees. This is actually the mean orientation
        of all stimuli in each bin.

  """
  with np.load(data_name) as dobj:
    data = dict(**dobj)
  resp = data['resp']
  stimuli = data['stimuli']

  if bin_width > 1:
    # Bin neural responses and stimuli
    bins = np.digitize(stimuli, np.arange(0, 360 + bin_width, bin_width))
    stimuli_binned = np.array([stimuli[bins == i].mean() for i in np.unique(bins)])
    resp_binned = np.array([resp[bins == i, :].mean(0) for i in np.unique(bins)])
  else:
    resp_binned = resp
    stimuli_binned = stimuli

  # only use stimuli <= 180
  resp_binned = resp_binned[stimuli_binned <= 180]
  stimuli_binned = stimuli_binned[stimuli_binned <= 180]

  stimuli_binned -= 90  # 0 means vertical, -ve means tilted left, +ve means tilted right

  # Return as torch.Tensor
  resp_tensor = torch.tensor(resp_binned, dtype=torch.float32)
  stimuli_tensor = torch.tensor(stimuli_binned, dtype=torch.float32).unsqueeze(1)  # add singleton dimension to make a column vector

  return resp_tensor, stimuli_tensor


def grating(angle, sf=1 / 28, res=0.1, patch=False):
  """Generate oriented grating stimulus

  Args:
    angle (float): orientation of grating (angle from vertical), in degrees
    sf (float): controls spatial frequency of the grating
    res (float): resolution of image. Smaller values will make the image
      smaller in terms of pixels. res=1.0 corresponds to 640 x 480 pixels.
    patch (boolean): set to True to make the grating a localized
      patch on the left side of the image. If False, then the
      grating occupies the full image.

  Returns:
    torch.Tensor: (res * 480) x (res * 640) pixel oriented grating image

  """

  angle = np.deg2rad(angle)  # transform to radians

  wpix, hpix = 640, 480  # width and height of image in pixels for res=1.0

  xx, yy = np.meshgrid(sf * np.arange(0, wpix * res) / res, sf * np.arange(0, hpix * res) / res)

  if patch:
    gratings = np.cos(xx * np.cos(angle + .1) + yy * np.sin(angle + .1))  # phase shift to make it better fit within patch
    gratings[gratings < 0] = 0
    gratings[gratings > 0] = 1
    xcent = gratings.shape[1] * .75
    ycent = gratings.shape[0] / 2
    xxc, yyc = np.meshgrid(np.arange(0, gratings.shape[1]), np.arange(0, gratings.shape[0]))
    icirc = ((xxc - xcent) ** 2 + (yyc - ycent) ** 2) ** 0.5 < wpix / 3 / 2 * res
    gratings[~icirc] = 0.5

  else:
    gratings = np.cos(xx * np.cos(angle) + yy * np.sin(angle))
    gratings[gratings < 0] = 0
    gratings[gratings > 0] = 1

  # Return torch tensor
  return torch.tensor(gratings, dtype=torch.float32)
```

```python
resp, stimuli = load_data(fname, bin_width=1)

imgs = [grating(angle, res=0.1) for angle in stimuli[:, 0].numpy()]
img = torch.stack(imgs)
img = img.view(img.shape[0], -1)

# Input Normalization
img = (img - img.min()) / (img.max() - img.min())

print(f"Generated Stimuli: {img.shape}")
```

## 3. Model Architecture
Implementation of a multi-layer Unsupervised Predictive Coding Network (PCN) utilizing local prediction error minimization.

```python
class PCNUnsupervised:
    def __init__(self, x_batch, layer_sizes, Tinfer, eta_infer, eta_learn):
        self.batch_size = x_batch.size(0)
        self.L = len(layer_sizes) - 1
        self.Tinfer = Tinfer
        self.Tlearn = self.batch_size
        self.eta_infer = eta_infer
        self.eta_learn = eta_learn

        self.x = [None] * (self.L + 1)
        self.x[0] = x_batch.clone().detach()

        self.W = [None] * self.L
        self.x_hat = [None] * self.L
        self.errors = [None] * self.L

        for l in range(self.L):
            w_tensor = torch.empty(layer_sizes[l], layer_sizes[l+1]).requires_grad_(True)
            torch.nn.init.xavier_uniform_(w_tensor)
            self.W[l] = w_tensor
            self.x[l+1] = (torch.randn(self.batch_size, layer_sizes[l+1]) * 0.1).requires_grad_(True)

    def relu(self, x):
        return torch.relu(x)

    def inference(self):
        for t in range(self.Tinfer):
            for l in range(1, self.L + 1):
                if self.x[l].grad is not None:
                    self.x[l].grad.zero_()

            total_energy = 0.0
            for l in range(self.L):
                # Prediction: X_{l+1} @ W_l.T
                self.x_hat[l] = self.relu(torch.matmul(self.x[l+1], self.W[l].t()))
                self.errors[l] = self.x[l] - self.x_hat[l]
                total_energy += 0.5 * torch.sum(self.errors[l] ** 2)

            for l in range(1, self.L + 1):
                self.x[l].retain_grad()

            total_energy.backward()

            with torch.no_grad():
                for l in range(1, self.L + 1):
                    if self.x[l].grad is not None:
                        self.x[l] -= self.eta_infer * self.x[l].grad
                        self.x[l].grad.zero_()
        return total_energy.item()

    def learn(self):
        final_energy = 0.0
        for t in range(self.Tlearn):
            for l in range(self.L):
                if self.W[l].grad is not None:
                    self.W[l].grad.zero_()

            total_energy = 0.0
            for l in range(self.L):
                self.x_hat[l] = self.relu(torch.matmul(self.x[l+1], self.W[l].t()))
                self.errors[l] = self.x[l] - self.x_hat[l]
                total_energy += 0.5 * torch.sum(self.errors[l] ** 2)

            total_energy.backward()

            with torch.no_grad():
                for l in range(self.L):
                    self.W[l] -= self.eta_learn * self.W[l].grad
                    self.W[l].grad.zero_()
            final_energy = total_energy.item()

        return final_energy
```

## 3.1 Supervised PCN Implementation
In this version, the top layer is constrained by the actual stimulus labels (orientations), and the error is backpropagated through the hierarchy to align representations with the task.

```python
class PCNSupervised:
    def __init__(self, x_batch, y_batch, layer_sizes, Tinfer, eta_infer, eta_learn):
        self.batch_size = x_batch.size(0)
        self.L = len(layer_sizes) - 1
        self.Tinfer = Tinfer
        self.Tlearn = self.batch_size
        self.eta_infer = eta_infer
        self.eta_learn = eta_learn

        self.x = [None] * (self.L + 1)
        self.x[0] = x_batch.clone().detach()
        # Top layer is constrained by labels
        self.y = y_batch.clone().detach()

        self.W = [None] * self.L
        self.x_hat = [None] * self.L
        self.errors = [None] * self.L

        for l in range(self.L):
            w_tensor = torch.empty(layer_sizes[l], layer_sizes[l+1]).requires_grad_(True)
            torch.nn.init.xavier_uniform_(w_tensor)
            self.W[l] = w_tensor
            # Initialize latents
            self.x[l+1] = (torch.randn(self.batch_size, layer_sizes[l+1]) * 0.1).requires_grad_(True)

    def relu(self, x):
        return torch.relu(x)

    def inference(self):
        # Fix the top layer to the labels during supervised inference/learning
        with torch.no_grad():
            self.x[self.L].copy_(self.y)

        for t in range(self.Tinfer):
            for l in range(1, self.L): # Only update intermediate layers
                if self.x[l].grad is not None:
                    self.x[l].grad.zero_()

            total_energy = 0.0
            for l in range(self.L):
                self.x_hat[l] = self.relu(torch.matmul(self.x[l+1], self.W[l].t()))
                self.errors[l] = self.x[l] - self.x_hat[l]
                total_energy += 0.5 * torch.sum(self.errors[l] ** 2)

            for l in range(1, self.L):
                self.x[l].retain_grad()

            total_energy.backward()

            with torch.no_grad():
                for l in range(1, self.L):
                    if self.x[l].grad is not None:
                        self.x[l] -= self.eta_infer * self.x[l].grad
                        self.x[l].grad.zero_()
        return total_energy.item()

    def learn(self):
        final_energy = 0.0
        # Clamp top layer
        with torch.no_grad():
            self.x[self.L].copy_(self.y)

        for l in range(self.L):
            if self.W[l].grad is not None:
                self.W[l].grad.zero_()

        total_energy = 0.0
        for l in range(self.L):
            self.x_hat[l] = self.relu(torch.matmul(self.x[l+1], self.W[l].t()))
            self.errors[l] = self.x[l] - self.x_hat[l]
            total_energy += 0.5 * torch.sum(self.errors[l] ** 2)

        total_energy.backward()

        with torch.no_grad():
            for l in range(self.L):
                self.W[l] -= self.eta_learn * self.W[l].grad
                self.W[l].grad.zero_()

        return total_energy.item()
```

```python
pcn = PCNUnsupervised(
    x_batch=img,
    layer_sizes=[3072, 256, 64],
    Tinfer=50,
    eta_infer=0.01,
    eta_learn=0.0002
)

for epoch in range(120):
    inf_energy = pcn.inference()
    learn_energy = pcn.learn()
    if (epoch + 1) % 10 == 0 or epoch == 0:
        print(f"Epoch {epoch + 1} | Inference Energy: {inf_energy:.4f} | Learning Energy: {learn_energy:.4f}")

# Save learned weights
weights = {f"W_{l}": pcn.W[l].detach().cpu() for l in range(pcn.L)}
torch.save(weights, "pcn_v1_weights.pt")
print("Training complete. Weights saved.")
```

### 4.1 Representational Similarity Analysis (RSA)
We calculate the correlation between the Representational Dissimilarity Matrices (RDMs) of our network layers and the V1 recordings.

```python
from scipy.stats import spearmanr
from scipy.spatial.distance import pdist, squareform

def rdm(matrix):
    if torch.is_tensor(matrix):
        m = matrix.detach().cpu().numpy()
    else:
        m = matrix

    if m.ndim > 2:
        m = m.reshape(m.shape[0], -1)

    return squareform(pdist(m, metric='correlation'))

# RDMs
rdm_v1  = rdm(resp)
rdm_l1  = rdm(acts[1])
rdm_l2  = rdm(acts[2])

n_stim = resp.shape[0]
idx = np.triu_indices(n_stim, k=1)

def rsa_score(rdm_a, rdm_b):
    r, p = spearmanr(rdm_a[idx], rdm_b[idx])
    return r, p

print(f"--- RSA Analysis (n={n_stim} stimuli) ---")
for l, rdm_pcn in [(1, rdm_l1), (2, rdm_l2)]:
    r, p = rsa_score(rdm_v1, rdm_pcn)
    print(f"Layer {l} — RSA r = {r:.3f}, p = {p:.4e}")

# Tuning curves
stim_1d = stimuli.cpu().numpy().flatten()
sort_idx = np.argsort(stim_1d)
orientations_sorted = stim_1d[sort_idx]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for i in range(3):
    unit_response = acts[1][:, i].detach().cpu().numpy().flatten()
    unit_response_sorted = unit_response[sort_idx]
    axes[i].plot(orientations_sorted, unit_response_sorted, color='green', linewidth=2)
    axes[i].set_title(f"Layer 1 - Unit {i}")
    axes[i].set_xlabel("Orientation (°)")
    axes[i].set_ylabel("Activation")
    axes[i].grid(True, alpha=0.3)

plt.suptitle("Tuning curves - Layer 1", fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
